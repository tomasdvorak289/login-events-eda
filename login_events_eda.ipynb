```python
# ============================================================
# PROJEKT: Login Events – Exploratory Data Analysis (EDA)
# ============================================================
#
# CÍL:
# Vytvořit jednoduchou analýzu 10 fiktivních přihlášení
# do počítačového systému pomocí pandas a Matplotlib.
#
# ÚKOL 1 – VYTVOŘENÍ DAT
#
# 1. Importuj knihovny pandas a matplotlib.pyplot.
# 2. Vytvoř slovník (dictionary) obsahující 10 přihlášení.
# 3. Každé přihlášení musí obsahovat:
#    - user: uživatelské jméno
#    - country: zemi přihlášení
#    - success: True (úspěšné) nebo False (neúspěšné)
# 4. Ve slovníku použij alespoň 3 různé uživatele
#    a alespoň 3 různé země.
# 5. Převeď slovník na pandas DataFrame.
# 6. Vypiš výslednou tabulku.


import pandas as pd
import matplotlib.pyplot as plt

logins = {
    'user': ['admin', 'tom', 'jirka', 'admin', 'tom',
             'jirka', 'admin', 'tom', 'jirka', 'admin'],

    'country': ['CZ', 'EN', 'DE', 'RU', 'CZ',
                'DE', 'CZ', 'EN', 'RU', 'CZ'],

    'success': [True, False, True, False, True,
                False, True, True, False, True]
}

df_logins = pd.DataFrame(logins)
print(df_logins)
```

```python

# ÚKOL 2 – PRVNÍ GRAF
#
# 1. Zjisti, kolik přihlášení pochází z jednotlivých zemí.
sub_df_logins = df_logins.loc[:,'country'].value_counts()
# 2. Vytvoř sloupcový graf pomocí Matplotlib.

plt.bar(x=sub_df_logins.index, height=sub_df_logins.values)

# 3. Přidej název grafu a popisky os.
plt.title('Výskyty přihlášení z jednotlivých zemí')
plt.xlabel('Země')
plt.ylabel('Počet pokusů o přihlášení')

plt.show()
```

```python
# ÚKOL 3 – DRUHÝ GRAF
#
# 1. Zjisti počet úspěšných a neúspěšných přihlášení.
# 2. Vytvoř druhý graf pomocí Matplotlib.
# 3. Přidej název grafu a popisky.
logins_stats = df_logins.loc[:, 'success'].value_counts()
print(logins_stats)
x_label_names = ['Úspěšné', 'Neúspěšné']
ordered_height = [logins_stats[True], logins_stats[False]]


plt.bar(x=x_label_names, height=ordered_height)
plt.title('Počty úspěšných nebo neúspěšných přihlášení')
plt.xlabel('Přihlášení bylo úspěšné/neúspěšné')
plt.ylabel('Počty jednotlivých příhlášení')
plt.show()

```
